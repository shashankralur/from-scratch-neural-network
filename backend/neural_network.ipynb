```python
def neuron(w, x):
    return w * x
```

```python
neuron(4, 5)
```

```python
def error(w,x,y):
    prediction = neuron(w, x)
    err = prediction - y
    return err
```

```python
error(4, 2, 6)
```

```python
def loss(w,x,y):
    err = error(w, x, y)
    return err ** 2
```

```python
loss(4, 2, 6)
```
